```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Breast Cancer Wisconsin (Diagnostic) Data Set.csv')
df.head()
```

```python
df.drop(columns=['id','Unnamed: 32'],inplace=True)
df.head()
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,1:], df.iloc[:,0],test_size=0.2, random_state=2)
```

```python
X_train.head()
```

```python
X_train.shape
```

```python
# Need to scale the data before training the model, as the features may have different ranges and scales. This can help improve the performance of many machine learning algorithms.
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
X_train.shape
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3)
```

```python
knn.fit(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = knn.predict(X_test)

accuracy_score(y_test, y_pred)
```

```python
scores = []

for i in range(1,16):
    
    knn = KNeighborsClassifier(n_neighbors=i)
    
    knn.fit(X_train,y_train)
    
    y_pred = knn.predict(X_test)

    scores.append(accuracy_score(y_test, y_pred))
    
```

```python
import matplotlib.pyplot as plt

plt.plot(range(1,16),scores)
```

```python
# define the function to plot decision boundaries

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from sklearn import neighbors, datasets
from sklearn.preprocessing import StandardScaler
from ipywidgets import interact, fixed

def load_data():
    cancer = datasets.load_breast_cancer()
    return cancer

def plot_decision_boundaries(n_neighbors, data, labels):
    h = .02
    cmap_light = ListedColormap(['orange', 'blue'])
    cmap_bold = ListedColormap(['darkorange', 'darkblue'])

    clf = neighbors.KNeighborsClassifier(n_neighbors)
    clf.fit(data, labels)

    x_min, x_max = data[:, 0].min() - 1, data[:, 0].max() + 1
    y_min, y_max = data[:, 1].min() - 1, data[:, 1].max() + 1

    xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))
    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])

    Z = Z.reshape(xx.shape)
    plt.figure(figsize=(8, 6))
    plt.pcolormesh(xx, yy, Z, cmap=cmap_light)

    plt.scatter(data[:, 0], data[:, 1], c=labels, cmap=cmap_bold, edgecolor='k', s=20)
    plt.xlim(xx.min(), xx.max())
    plt.ylim(yy.min(), yy.max())
    plt.title(f'2-Class classification (k = {n_neighbors})')
    plt.show()

cancer = load_data()

# Use only the first two features and standardize them.
X = StandardScaler().fit_transform(cancer.data[:, :2])
y = cancer.target

# Interactive widget
interact(plot_decision_boundaries, n_neighbors=(1, 20), data=fixed(X), labels=fixed(y));
```

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn import datasets
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

from mlxtend.plotting import plot_decision_regions


# Load Breast Cancer dataset
cancer = datasets.load_breast_cancer()

# Use only first two features
X = cancer.data[:, :2]
y = cancer.target

# Standardize features
X = StandardScaler().fit_transform(X)


# Define function to plot decision boundaries
def plot_knn_decision_boundary(n_neighbors):

    # Create KNN classifier
    knn = KNeighborsClassifier(n_neighbors=n_neighbors)

    # Fit model
    knn.fit(X, y)

    # Create figure
    plt.figure(figsize=(8, 6))

    # Plot decision regions
    plot_decision_regions(
        X=X,
        y=y,
        clf=knn,
        legend=2
    )

    plt.xlabel(cancer.feature_names[0])
    plt.ylabel(cancer.feature_names[1])

    plt.title(
        f"KNN Decision Boundary (k = {n_neighbors})"
    )

    plt.show()


# Try different values of K
plot_knn_decision_boundary(5)
```
