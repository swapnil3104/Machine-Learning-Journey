# Python Tuples

### What is a Tuple?

A **tuple** is an ordered, immutable (unchangeable) collection in Python. Tuples can store multiple items of different data types.

### Characteristics of Tuples

- Ordered
- Immutable (cannot be modified after creation)
- Allows duplicate values
- Can store different data types
- Faster than lists for read-only data



````python
fruits = ("apple", "banana", "cherry")
print(fruits)
````

````python
print(type(fruits))  # Output: <class 'tuple'>
````

## Python - Access Tuple Items

Tuple items are accessed using **index numbers**. Python uses **zero-based indexing**, which means the first item has an index of `0`.

---

### Access Items by Index

You can access a specific item by referring to its index.

#### Example

```python
fruits = ("apple", "banana", "cherry")

print(fruits[0])
print(fruits[1])
print(fruits[2])
```

**Output:**

```
apple
banana
cherry
```

---

### Negative Indexing

Negative indexing starts from the end of the tuple.

- `-1` → Last item
- `-2` → Second last item
- `-3` → Third last item

#### Example

```python
fruits = ("apple", "banana", "cherry", "mango")

print(fruits[-1])
print(fruits[-2])
print(fruits[-3])
```

**Output:**

```
mango
cherry
banana
```

---

### Access a Range of Items (Slicing)

Use the slicing syntax to access a range of items.

**Syntax:**

```python
tuple[start:stop]
```

- `start` is included.
- `stop` is excluded.

#### Example

```python
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[1:4])
```

**Output:**

```
(20, 30, 40)
```

---

### Slice from the Beginning

If the start index is omitted, slicing begins from the first item.

#### Example

```python
numbers = (10, 20, 30, 40, 50)

print(numbers[:3])
```

**Output:**

```
(10, 20, 30)
```

---

### Slice to the End

If the stop index is omitted, slicing continues to the last item.

#### Example

```python
numbers = (10, 20, 30, 40, 50)

print(numbers[2:])
```

**Output:**

```
(30, 40, 50)
```

---

### Slice Using Negative Indexes

You can also use negative indexes with slicing.

#### Example

```python
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[-4:-1])
```

**Output:**

```
(30, 40, 50)
```

---

### Access Nested Tuple Items

A tuple can contain another tuple. Use multiple indexes to access nested items.

#### Example

```python
students = (
    ("John", 101),
    ("Emma", 102),
    ("David", 103)
)

print(students[0][0])
print(students[1][1])
```

**Output:**

```
John
102
```

---

### Loop Through Tuple Items

You can access each item using a `for` loop.

#### Example

```python
fruits = ("apple", "banana", "cherry")

for fruit in fruits:
    print(fruit)
```

**Output:**

```
apple
banana
cherry
```

---

### Check if an Item Exists

Use the `in` operator to check whether an item exists in a tuple.

#### Example

```python
fruits = ("apple", "banana", "cherry")

print("banana" in fruits)
print("grapes" in fruits)
```

**Output:**

```
True
False
```

---

### Summary

- Access tuple items using indexes.
- Indexing starts from `0`.
- Use negative indexes to access items from the end.
- Use slicing to access a range of items.
- Nested tuples require multiple indexes.
- Use loops to access all items.
- Use the `in` operator to check if an item exists.

````python
Bike = ("Honda", "Yamaha", "Suzuki")
print(Bike)
````

## Python - Update Tuples

Tuples are **immutable**, which means their values **cannot be changed** after they are created.

If you need to update a tuple, you must first convert it to a list, make the changes, and then convert it back to a tuple.

---

### Change Tuple Values

Since tuples are immutable, you cannot directly modify an item.

#### Incorrect Example

```python
fruits = ("apple", "banana", "cherry")

fruits[1] = "mango"
```

**Output:**

```
TypeError: 'tuple' object does not support item assignment
```

---

### Update a Tuple by Converting It to a List

Convert the tuple to a list, modify the list, and convert it back to a tuple.

#### Example

```python
fruits = ("apple", "banana", "cherry")

temp = list(fruits)
temp[1] = "mango"
fruits = tuple(temp)

print(fruits)
```

**Output:**

```
('apple', 'mango', 'cherry')
```

---

### Add an Item to a Tuple

Since tuples cannot be modified directly, convert the tuple to a list, add the item, and convert it back.

#### Example

```python
fruits = ("apple", "banana", "cherry")

temp = list(fruits)
temp.append("orange")
fruits = tuple(temp)

print(fruits)
```

**Output:**

```
('apple', 'banana', 'cherry', 'orange')
```

---

### Insert an Item into a Tuple

Use the `insert()` method after converting the tuple to a list.

#### Example

```python
fruits = ("apple", "banana", "cherry")

temp = list(fruits)
temp.insert(1, "mango")
fruits = tuple(temp)

print(fruits)
```

**Output:**

```
('apple', 'mango', 'banana', 'cherry')
```

---

### Remove an Item from a Tuple

Convert the tuple to a list, remove the item, and convert it back.

#### Example

```python
fruits = ("apple", "banana", "cherry")

temp = list(fruits)
temp.remove("banana")
fruits = tuple(temp)

print(fruits)
```

**Output:**

```
('apple', 'cherry')
```

---

### Delete an Entire Tuple

You cannot delete individual items directly, but you can delete the entire tuple using the `del` keyword.

#### Example

```python
fruits = ("apple", "banana", "cherry")

del fruits

print(fruits)
```

**Output:**

```
NameError: name 'fruits' is not defined
```

---

### Create a New Tuple by Concatenation

You can create a new tuple by joining two tuples.

#### Example

```python
fruits = ("apple", "banana")
new_items = ("cherry", "orange")

fruits = fruits + new_items

print(fruits)
```

**Output:**

```
('apple', 'banana', 'cherry', 'orange')
```

---

### Summary

- Tuples are **immutable**, so their values cannot be changed directly.
- Convert a tuple to a list to update, add, insert, or remove items.
- Convert the list back to a tuple after making changes.
- Use the `+` operator to create a new tuple by combining tuples.
- Use `del` to delete an entire tuple.
```

## Python - Unpack Tuples

Tuple unpacking allows you to assign the values of a tuple to multiple variables in a single statement.

---

### Unpacking a Tuple

When unpacking, the number of variables must match the number of items in the tuple.

#### Example

```python
fruits = ("apple", "banana", "cherry")

fruit1, fruit2, fruit3 = fruits

print(fruit1)
print(fruit2)
print(fruit3)
```

**Output:**

```
apple
banana
cherry
```

---

### Unpacking Numbers

You can unpack numeric values into variables.

#### Example

```python
numbers = (10, 20, 30)

a, b, c = numbers

print(a)
print(b)
print(c)
```

**Output:**

```
10
20
30
```

---

### Unpacking Different Data Types

A tuple can contain different data types, and each value can be unpacked into a separate variable.

#### Example

```python
person = ("Alice", 22, "Engineer", True)

name, age, profession, employed = person

print(name)
print(age)
print(profession)
print(employed)
```

**Output:**

```
Alice
22
Engineer
True
```

---

### Using the Asterisk (*) Operator

If there are more items than variables, use the `*` operator to collect the remaining items into a list.

#### Example

```python
fruits = ("apple", "banana", "cherry", "mango", "orange")

first, *middle, last = fruits

print(first)
print(middle)
print(last)
```

**Output:**

```
apple
['banana', 'cherry', 'mango']
orange
```

---

### Collect Remaining Items at the End

The `*` operator can also collect the remaining items at the end.

#### Example

```python
fruits = ("apple", "banana", "cherry", "mango")

first, *remaining = fruits

print(first)
print(remaining)
```

**Output:**

```
apple
['banana', 'cherry', 'mango']
```

---

### Collect Remaining Items at the Beginning

The `*` operator can collect items at the beginning.

#### Example

```python
fruits = ("apple", "banana", "cherry", "mango")

*remaining, last = fruits

print(remaining)
print(last)
```

**Output:**

```
['apple', 'banana', 'cherry']
mango
```

---

### Unpacking Nested Tuples

You can unpack tuples that contain other tuples.

#### Example

```python
student = ("John", (85, 90, 95))

name, (math, science, english) = student

print(name)
print(math)
print(science)
print(english)
```

**Output:**

```
John
85
90
95
```

---

### Incorrect Unpacking

If the number of variables does not match the number of tuple items, Python raises a `ValueError`.

#### Example

```python
fruits = ("apple", "banana", "cherry")

a, b = fruits
```

**Output:**

```
ValueError: too many values to unpack (expected 2)
```

---

### Summary

- Tuple unpacking assigns tuple values to variables.
- The number of variables should match the number of tuple items.
- Use the `*` operator to collect multiple values into a list.
- The `*` operator can appear at the beginning, middle, or end.
- Nested tuples can also be unpacked.
- Incorrect unpacking raises a `ValueError`.

## Python - Loop Tuples

You can loop through the items of a tuple using different looping techniques such as the `for` loop, `while` loop, and by using indexes.

---

### Loop Through a Tuple

The easiest way to loop through a tuple is by using a `for` loop.

#### Example

```python
fruits = ("apple", "banana", "cherry")

for fruit in fruits:
    print(fruit)
```

**Output:**

```
apple
banana
cherry
```

---

### Loop Through the Index Numbers

Use the `range()` function along with `len()` to access tuple items using their index.

#### Example

```python
fruits = ("apple", "banana", "cherry")

for i in range(len(fruits)):
    print(fruits[i])
```

**Output:**

```
apple
banana
cherry
```

---

### Loop Using a While Loop

You can also use a `while` loop to iterate through a tuple.

#### Example

```python
fruits = ("apple", "banana", "cherry")

i = 0

while i < len(fruits):
    print(fruits[i])
    i += 1
```

**Output:**

```
apple
banana
cherry
```

---

### Loop Through a Tuple of Numbers

#### Example

```python
numbers = (10, 20, 30, 40, 50)

for number in numbers:
    print(number)
```

**Output:**

```
10
20
30
40
50
```

---

### Loop with `enumerate()`

The `enumerate()` function returns both the index and the value during iteration.

#### Example

```python
fruits = ("apple", "banana", "cherry")

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

**Output:**

```
0 apple
1 banana
2 cherry
```

---

### Loop Through a Nested Tuple

You can use nested loops to iterate through nested tuples.

#### Example

```python
students = (
    ("John", 101),
    ("Emma", 102),
    ("David", 103)
)

for student in students:
    for item in student:
        print(item)
```

**Output:**

```
John
101
Emma
102
David
103
```

---

### Summary

- Use a `for` loop to iterate through tuple items.
- Use `range()` and `len()` to loop using index numbers.
- Use a `while` loop for manual iteration.
- Use `enumerate()` to get both the index and value.
- Use nested loops to iterate through nested tuples.
```

## Python - Join Tuples

You can join two or more tuples using the **`+` (concatenation)** operator. Joining tuples creates a **new tuple** because tuples are immutable.

---

### Join Two Tuples

Use the `+` operator to combine two tuples.

#### Example

```python
tuple1 = ("apple", "banana")
tuple2 = ("cherry", "mango")

result = tuple1 + tuple2

print(result)
```

**Output:**

```
('apple', 'banana', 'cherry', 'mango')
```

---

### Join Multiple Tuples

You can join more than two tuples using the `+` operator.

#### Example

```python
tuple1 = (1, 2)
tuple2 = (3, 4)
tuple3 = (5, 6)

result = tuple1 + tuple2 + tuple3

print(result)
```

**Output:**

```
(1, 2, 3, 4, 5, 6)
```

---

### Join Tuples with Different Data Types

A tuple can contain different data types, and they can be joined together.

#### Example

```python
tuple1 = ("Alice", 25)
tuple2 = ("Engineer", True)

result = tuple1 + tuple2

print(result)
```

**Output:**

```
('Alice', 25, 'Engineer', True)
```

---

### Join an Empty Tuple

Joining a tuple with an empty tuple returns the original tuple.

#### Example

```python
numbers = (10, 20, 30)
empty = ()

result = numbers + empty

print(result)
```

**Output:**

```
(10, 20, 30)
```

---

### Join Tuples Using the `+=` Operator

The `+=` operator creates a new tuple by joining another tuple.

#### Example

```python
fruits = ("apple", "banana")

fruits += ("cherry",)

print(fruits)
```

**Output:**

```
('apple', 'banana', 'cherry')
```

> **Note:** When adding a single item, include a comma after the value to create a tuple.

---

### Difference Between `+` and `+=`

| Operator | Description |
|----------|-------------|
| `+` | Creates a new tuple by joining two tuples. |
| `+=` | Reassigns the variable to a new joined tuple. |

---

### Summary

- Use the `+` operator to join tuples.
- Joining tuples creates a new tuple because tuples are immutable.
- You can join two or more tuples.
- Tuples with different data types can also be joined.
- Use `+=` to append another tuple by creating a new tuple.
- A single-item tuple must include a trailing comma.

## Python - Tuple Methods

Python tuples have only **two built-in methods** because tuples are **immutable**. These methods are:

- `count()`
- `index()`

---

### Tuple Methods Overview

| Method | Description |
|---------|-------------|
| `count()` | Returns the number of times a specified value appears in the tuple. |
| `index()` | Returns the index of the first occurrence of a specified value. |

---

### `count()` Method

The `count()` method returns the number of times a specified value appears in the tuple.

**Syntax:**

```python
tuple.count(value)
```

#### Example

```python
numbers = (1, 2, 2, 3, 2, 4)

print(numbers.count(2))
```

**Output:**

```
3
```

---

### `count()` with Strings

#### Example

```python
fruits = ("apple", "banana", "apple", "cherry", "apple")

print(fruits.count("apple"))
```

**Output:**

```
3
```

---

### `count()` When the Value Does Not Exist

If the value is not found, `count()` returns `0`.

#### Example

```python
numbers = (10, 20, 30)

print(numbers.count(50))
```

**Output:**

```
0
```

---

### `index()` Method

The `index()` method returns the index of the **first occurrence** of a specified value.

**Syntax:**

```python
tuple.index(value)
```

#### Example

```python
fruits = ("apple", "banana", "cherry", "banana")

print(fruits.index("banana"))
```

**Output:**

```
1
```

---

### `index()` with Numbers

#### Example

```python
numbers = (10, 20, 30, 40)

print(numbers.index(30))
```

**Output:**

```
2
```

---

### `index()` When the Value Does Not Exist

If the specified value is not found, Python raises a `ValueError`.

#### Example

```python
numbers = (10, 20, 30)

print(numbers.index(50))
```

**Output:**

```
ValueError: tuple.index(x): x not in tuple
```

---

### Using `count()` and `index()` Together

#### Example

```python
numbers = (5, 10, 15, 10, 20)

print("Count:", numbers.count(10))
print("First Index:", numbers.index(10))
```

**Output:**

```
Count: 2
First Index: 1
```

---

### Summary

- Tuples have only **two built-in methods**.
- `count()` returns the number of occurrences of a value.
- `index()` returns the index of the first occurrence of a value.
- `count()` returns `0` if the value is not found.
- `index()` raises a `ValueError` if the value is not found.
- Both methods work with numbers, strings, and other data types stored in tuples.
