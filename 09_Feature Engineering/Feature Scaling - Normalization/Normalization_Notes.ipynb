# Feature Scaling

## Normalization (Min-Max Scaling)

Normalization is a **feature scaling** technique used in Machine Learning to transform numerical data into a fixed range, usually **0 to 1**. It ensures that all features have the same scale without changing the relationships between the original values.

Unlike standardization, normalization always maps the minimum value of a feature to **0** and the maximum value to **1**. All other values are transformed proportionally between these limits.

Normalization is also known as **Min-Max Scaling** because it uses the minimum and maximum values of the dataset during the transformation.

---

## Why Do We Need Normalization?

In real-world datasets, different features often have completely different ranges.

For example:

| Feature | Range |
|----------|-------|
| Age | 18 – 60 |
| Salary | ₹20,000 – ₹2,00,000 |
| Height (cm) | 140 – 190 |

If these features are used directly in machine learning algorithms, features with larger numerical values (such as salary) can dominate features with smaller values (such as age).

Normalization solves this problem by converting every feature into the **same range (0 to 1)** so that each feature contributes equally during model training.

---

## What Does Normalization Do?

After normalization:

- The **minimum value** becomes **0**.
- The **maximum value** becomes **1**.
- Every other value lies between **0 and 1**.
- The relative distance between data points remains the same.

For example,

Original values:

```
20   40   60   80   100
```

Normalized values:

```
0   0.25   0.50   0.75   1
```

---

## Formula

The mathematical formula for Min-Max Normalization is:

\[
X_{norm}=\frac{X-X_{min}}{X_{max}-X_{min}}
\]

Where:

| Symbol | Meaning |
|---------|----------|
| **X** | Original value |
| **Xmin** | Minimum value in the dataset |
| **Xmax** | Maximum value in the dataset |
| **Xnorm** | Normalized value |

---


## Advantages of Normalization

- Scales all features to the same range (0 to 1).
- Easy to understand and interpret.
- Prevents features with larger values from dominating smaller-valued features.
- Improves the performance of distance-based algorithms.
- Useful when the data does not follow a normal distribution.

---

## Disadvantages of Normalization

- Highly sensitive to outliers.
- A single extreme value can compress the remaining data into a very small range.
- If new data contains values outside the original minimum or maximum, normalization may produce values outside the range 0 to 1 unless the scaler is refitted.

---

## Algorithms That Require Normalization

Normalization is commonly recommended for:

- K-Nearest Neighbors (KNN)
- K-Means Clustering
- Neural Networks
- Support Vector Machine (SVM)
- Principal Component Analysis (PCA)
- Gradient Descent-based algorithms

---

## Algorithms That Usually Do Not Require Normalization

Tree-based algorithms generally do not require feature scaling because they split data based on feature values rather than distances.

Examples include:

- Decision Tree
- Random Forest
- XGBoost
- LightGBM
- CatBoost

---

## Normalization vs Standardization

| Feature | Normalization | Standardization |
|----------|---------------|-----------------|
| Formula | (X − Min) / (Max − Min) | (X − μ) / σ |
| Output Range | 0 to 1 | Not fixed |
| Mean | Not fixed | 0 |
| Standard Deviation | Not fixed | 1 |
| Sensitive to Outliers | Very High | Moderate |
| Best Used For | Bounded data and distance-based models | Normally distributed data |

---

## When Should You Use Normalization?

Normalization is preferred when:

- Features have different scales.
- The algorithm uses distance calculations.
- You want all values between **0 and 1**.
- The dataset does not necessarily follow a normal distribution.
- Neural networks are being trained.

---

## Key Points to Remember

- Normalization is also called **Min-Max Scaling**.
- It converts data into a fixed range, usually **0 to 1**.
- The smallest value becomes **0**, and the largest becomes **1**.
- It preserves the relative ordering of data.
- It is commonly used before training distance-based machine learning models.
- Normalization is sensitive to outliers because it depends on the minimum and maximum values of the dataset.

---

````python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Sample Data
df = pd.DataFrame({
    "Marks": [40, 50, 60, 70, 80]
})

# Create MinMaxScaler object
scaler = MinMaxScaler()

# Apply Normalization
df["Normalized_Marks"] = scaler.fit_transform(df[["Marks"]])

print(df)
````
