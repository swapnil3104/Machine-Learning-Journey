# Feature Scaling

## Standardization (Z-Score Standardization)

Standardization is one of the most commonly used **feature scaling** techniques in Machine Learning. It transforms numerical data so that each feature has a **mean of 0** and a **standard deviation of 1**. This process ensures that all features contribute equally to the learning process, regardless of their original scale.

Unlike normalization, standardization does **not** restrict values to a fixed range such as 0 to 1. Instead, it measures how far each data point is from the mean in terms of standard deviations. Therefore, the transformed values can be **negative, zero, or positive**.

---

## Why Do We Need Standardization?

Real-world datasets often contain features with different units and scales. For example:

| Feature | Range |
|----------|-------|
| Age | 18 – 60 |
| Salary | ₹20,000 – ₹2,00,000 |
| Years of Experience | 0 – 40 |

Machine learning algorithms that calculate distances or optimize gradients can become biased toward features with larger values.

For example, consider these two features:

- Age = 25
- Salary = ₹150,000

Since salary values are much larger than age values, an algorithm may treat salary as more important simply because of its larger magnitude. Standardization solves this problem by converting all features to the same scale.

---

## What Does Standardization Do?

After standardization:

- Mean of the feature becomes **0**
- Standard deviation becomes **1**
- Values above the mean become positive.
- Values below the mean become negative.
- Values equal to the mean become **0**.

This transformation makes the learning algorithm focus on the relationship between features rather than their magnitude.

---

## Formula

The mathematical formula for standardization is:

\[
Z = \frac{X - \mu}{\sigma}
\]

Where:

| Symbol | Meaning |
|---------|----------|
| **Z** | Standardized value (Z-score) |
| **X** | Original value |
| **μ (Mu)** | Mean of the feature |
| **σ (Sigma)** | Standard deviation of the feature |

---

## Advantages of Standardization

- Places all numerical features on a common scale.
- Improves the performance of gradient-based optimization algorithms.
- Speeds up model training.
- Prevents features with larger values from dominating smaller-valued features.
- Works well when data approximately follows a normal (Gaussian) distribution.
- Essential for many distance-based algorithms.

---

## Disadvantages of Standardization

- Does not restrict values to a fixed range.
- Sensitive to extreme outliers because the mean and standard deviation are affected by them.
- The transformed values may be less intuitive to interpret than the original values.

---

## Algorithms That Require Standardization

Standardization is highly recommended for:

- Linear Regression
- Logistic Regression
- Support Vector Machine (SVM)
- K-Nearest Neighbors (KNN)
- K-Means Clustering
- Principal Component Analysis (PCA)
- Neural Networks
- Gradient Descent-based algorithms

---

## Algorithms That Usually Do Not Require Standardization

Tree-based models are generally unaffected by feature scaling because they split data based on feature values rather than distances.

Examples include:

- Decision Tree
- Random Forest
- XGBoost
- LightGBM
- CatBoost

---

## Standardization vs Normalization

| Feature | Standardization | Normalization |
|----------|-----------------|---------------|
| Formula | (X − μ) / σ | (X − Min) / (Max − Min) |
| Output Range | Not fixed | 0 to 1 |
| Mean | 0 | Not fixed |
| Standard Deviation | 1 | Not fixed |
| Sensitive to Outliers | Yes | Yes (more sensitive) |
| Best Used For | Gaussian-distributed data | Data with known minimum and maximum |

---

## Key Points to Remember

- Standardization is also called **Z-Score Standardization**.
- It transforms features so that their **mean becomes 0** and **standard deviation becomes 1**.
- Values can be negative, zero, or positive.
- It is widely used before training many machine learning models.
- It helps algorithms converge faster and improves model accuracy.
- Standardization is one of the most important preprocessing techniques in machine learning.

---

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Sample Data
df = pd.DataFrame({
    "Marks": [40, 50, 60, 70, 80]
})

# Create StandardScaler object
scaler = StandardScaler()

# Apply Standardization
df["Standardized_Marks"] = scaler.fit_transform(df[["Marks"]])

print(df)
```
